## Tipos de datos

Python infiere el tipo de cada objeto. Los cuatro básicos son entero, decimal, texto y booleano. `type()` devuelve cuál es.

Vale la pena mirar el tipo cuando algo no funciona como esperaban, porque muchos errores vienen de una variable que quedó guardada como texto cuando debería ser número. Al leer una base con una columna que tiene una celda con un guión o un `n/d`, pandas la trae entera como texto y cualquier promedio falla.

```python
edad = 34
salario = 1250.75
provincia = "Buenos Aires"
ocupado = True

for x in (edad, salario, provincia, ocupado):
    print(x, type(x))
```

### Operaciones aritméticas

La división con `/` siempre devuelve decimal, aunque los dos operandos sean enteros. Y la potencia es `**`, no `^`, que en Python significa otra cosa (un operador lógico bit a bit) y por eso no da error, simplemente da un resultado raro.

```python
print(7 / 2)
print(7 // 2)    # división entera, descarta el resto
print(7 % 2)     # resto
print(7 ** 2)    # NO se usa ^
print(7 ^ 2)     # esto no es la potencia
```

### Texto

Los strings se concatenan con `+` y se repiten con `*`. Para armar texto que incluya valores conviene usar f-strings, que van con una `f` antes de las comillas y permiten meter cualquier expresión entre llaves. Los dos puntos adentro de las llaves controlan el formato, que es cómodo para imprimir resultados con la cantidad de decimales que uno quiere.

```python
nombre = "Buenos Aires"
print("Provincia: " + nombre)
print("-" * 30)

print(f"En {nombre} el salario es {salario:.2f}")
print(f"Y en miles, {salario/1000:.1f}")
```

### Booleanos y operadores lógicos

Las comparaciones devuelven `True` o `False`, y se combinan con `and`, `or` y `not`. Esto es exactamente lo que después vamos a usar para filtrar bases, así que conviene tenerlo cómodo.

`True` y `False` valen 1 y 0, así que sumar un vector de booleanos cuenta cuántos casos cumplen la condición. Ese es el atajo estándar para calcular una proporción.

```python
print(edad > 30)
print(edad > 30 and salario > 1000)
print(not ocupado)

# la suma de booleanos cuenta cuántos son True
print(sum([True, False, True, True]))
```

## Colecciones

Python trae cuatro formas de guardar varios valores juntos. En este curso usamos sobre todo listas y diccionarios.

| | Ordenada | Modificable | Permite repetidos |
|---|---|---|---|
| Lista `[ ]` | sí | sí | sí |
| Tupla `( )` | sí | no | sí |
| Set `{ }` | no | sí | no |
| Diccionario `{clave: valor}` | sí | sí | claves únicas |

### Listas

Los índices arrancan en 0, no en 1. El rango `a:b` incluye `a` y excluye `b`, así que `[1:3]` devuelve dos elementos, no tres. Los índices negativos cuentan desde el final, y `-1` es el último, que es la forma más limpia de pedirlo sin calcular el largo.

```python
salarios = [850, 1200, 1500, 980, 2100]

print(salarios[0])
print(salarios[-1])
print(salarios[1:3])
print(len(salarios))
```

```python
salarios.append(1750)
salarios.sort()
print(salarios)
print(sum(salarios) / len(salarios))
```

Cuando asignan una lista a otro nombr no se copia nada: los dos nombres apuntan al mismo objeto en memoria. Si modifican uno, el otro cambia también. Para llevarse una copia independiente hay que pedirla con `.copy()`.

Esto pasa con listas, diccionarios y también con DataFrames, y es el origen de una buena parte de los resultados inexplicables.

```python
a = [1, 2, 3]
b = a
b.append(99)
print(a)          # a también cambió

c = a.copy()
c.append(100)
print(c)
print(a)          # esta vez no
```

### Diccionarios

Guardan pares de clave y valor. Se usan para etiquetar cosas, y más adelante en el curso van a aparecer al definir los hiperparámetros de un modelo, que se pasan como diccionario.

```python
hogar = {"provincia": "Buenos Aires", "miembros": 4, "ingreso": 320000}

print(hogar["provincia"])
print(list(hogar.keys()))

hogar["pobre"] = hogar["ingreso"] < 400000
print(hogar)
```

### Tuplas y sets

Las tuplas son listas que no se pueden modificar, y se usan para cosas que no deberían cambiar, como un par de coordenadas. Los sets no tienen orden y eliminan repetidos, así que la forma rápida de ver qué valores distintos toma una variable es meterla en un set.

```python
coords = (-34.6, -58.4)
regiones = {"NOA", "NEA", "GBA", "NOA", "Cuyo"}
print(regiones)
```

## Control de flujo

En Python la indentación no es cuestión de estilo: es lo que define qué está adentro del bloque. Cuatro espacios es la convención, y mezclar espacios con tabulaciones da error.

```python
for s in salarios:
    if s < 1000:
        print(s, "bajo")
    elif s < 1800:
        print(s, "medio")
    else:
        print(s, "alto")
```

Las *list comprehensions* construyen una lista en una línea. Conviene reconocerlas aunque prefieran escribir el bucle largo.

```python
cuadrados = [s ** 2 for s in salarios]
altos = [s for s in salarios if s > 1200]
print(altos)
```

## Funciones

Se definen con `def` y devuelven con `return`. El texto entre comillas triples es el *docstring*, la documentación de la función, y es lo que aparece cuando alguien escribe `help(nombre)` o pone el signo de pregunta al lado. Escribirlo cuesta diez segundos y es lo que hace que el código sirva dentro de un mes.

Poner en funciones lo que se repite no es una cuestión de elegancia. Si la misma limpieza está copiada en cinco lugares y hay que corregir un criterio, hay cinco lugares donde equivocarse.

```python
def tasa_de_pobreza(ingresos, linea):
    """Proporción de ingresos por debajo de la línea."""
    pobres = [i for i in ingresos if i < linea]
    return len(pobres) / len(ingresos)

print(tasa_de_pobreza(salarios, 1200))
help(tasa_de_pobreza)
```

## numpy

`numpy` trabaja con vectores y matrices. La diferencia práctica con una lista es que las operaciones se aplican a todos los elementos de una vez, sin escribir un bucle. Eso se llama vectorización, y además de ser más corto es órdenes de magnitud más rápido, porque la operación se ejecuta en código compilado.

Esto importa de verdad en este curso. Cuando lleguemos a cross validation vamos a repetir una estimación miles de veces, y ahí la diferencia entre código vectorizado y bucles en Python es la diferencia entre segundos y minutos.

```python
x = np.array(salarios)      #array: vector o matriz de números, todos del mismo tipo, con una forma definida

print(x * 1.15)              # el 15% se aplica a todos, sin bucle
print(x.mean(), x.std())
print(np.log(x).round(3))
```

## pandas

`pandas` El objeto central es el **DataFrame**: una tabla con nombres de columna y un índice de filas. Es el equivalente a la base de Stata o al `data.frame` de R.

Para que el notebook corra sin depender de ningún archivo, simulamos una base. La semilla fija hace que a todos les salga exactamente lo mismo, que es la práctica estándar para que un resultado con componente aleatorio sea reproducible. Sin semilla, dos corridas del mismo código dan números distintos y no hay forma de verificar nada.

El salario lo generamos con una relación **conocida**: 25000 por año de educación, 3000 por año de edad y 60000 menos si es mujer. Y, como pasa en los datos reales, las mujeres de esta muestra tienen en promedio un año más de educación que los varones. 

```python
rng = np.random.default_rng(2026)   # semilla fija
n = 500                             # tamaño de la muestra simulada

mujer = rng.integers(0, 2, n)       # dummy de género: n valores del 0 al 2 sin incluir el 2

datos = pd.DataFrame({
    "edad": rng.integers(18, 65, n),
    # las mujeres de la muestra tienen en promedio un año más de educación
    "educacion": rng.integers(7, 17, n) + mujer,
    "region": rng.choice(["GBA", "Pampeana", "NOA", "NEA", "Cuyo"], n),
    "mujer": mujer,
})

datos["salario"] = (300000
                    + 25000 * datos["educacion"]
                    + 3000 * datos["edad"]
                    - 60000 * datos["mujer"]
                    + rng.normal(0, 80000, n)).round(0)

datos.head()
```

### Seleccionar y filtrar

`df["col"]` devuelve una columna. `df.loc[filas, columnas]` selecciona por nombre y `df.iloc[...]` por posición.

Los filtros se escriben como condiciones booleanas, igual que las de más arriba, pero con dos diferencias que hay que respetar. Se usan `&` y `|` en lugar de `and` y `or`, porque la comparación se hace elemento por elemento. Y cada condición va entre paréntesis, porque `&` tiene más precedencia que `>` y sin los paréntesis el error es incomprensible.

```python
print(datos["salario"].mean().round(0))

datos.loc[:, ["edad", "educacion", "salario"]].head()
```

```python
jovenes = datos[datos["edad"] < 30]
print(len(jovenes))

# cada condición entre paréntesis, y & en lugar de and
sel = datos[(datos["edad"] < 30) & (datos["educacion"] >= 12)]
print(len(sel))
```

Si filtran una base y después le escriben una columna al resultado, pandas avisa con un `SettingWithCopyWarning`. El motivo es que no está definido si el filtro devolvió una vista sobre la base original o una copia independiente, así que el cambio puede propagarse o no.

La solución es pedir la copia explícitamente con `.copy()` cuando la intención es trabajar sobre un subconjunto aparte. No es una advertencia para ignorar.

```python
sub = datos[datos["region"] == "GBA"].copy()
sub["salario_miles"] = sub["salario"] / 1000
sub.head(3)
```

### Crear columnas y agrupar

Crear una columna es asignarle a un nombre que todavía no existe. Las operaciones son vectorizadas, así que se escriben sobre la columna entera.

`groupby` es el equivalente a `bysort` de Stata o a `group_by` de dplyr. Se agrupa por una o varias columnas y se aplica una función a las demás. Cuando se agrupa siempre conviene pedir el `count` junto con la media, porque un promedio calculado sobre tres observaciones no dice nada y en la tabla se ve igual de prolijo que uno calculado sobre mil.

```python
datos["log_salario"] = np.log(datos["salario"])

# astype(int) convierte el booleano en una dummy 0/1
datos["universitario"] = (datos["educacion"] >= 16).astype(int)

datos.groupby("region")["salario"].agg(["mean", "count"]).round(0)
```

```python
datos.groupby(["region", "universitario"])["salario"].mean().round(0)
```

### Una primera vuelta sobre lo que vimos en la parte teórica

Generamos el salario restando 60000 a las mujeres. Si comparamos los promedios por sexo,
¿recuperamos ese número?

```python
brecha = (datos.loc[datos["mujer"] == 0, "salario"].mean()
          - datos.loc[datos["mujer"] == 1, "salario"].mean())

print(f"Diferencia de medias simple:  {brecha:>9,.0f}")
print(f"Educación promedio, varones:  {datos.loc[datos['mujer']==0,'educacion'].mean():>9.2f}")
print(f"Educación promedio, mujeres:  {datos.loc[datos['mujer']==1,'educacion'].mean():>9.2f}")
print(f"Coeficiente que pusimos:      {-60000:>9,.0f}")
```

La diferencia da bastante menos que los 60000 que pusimos, y la razón no es el ruido.

Las mujeres de esta muestra tienen en promedio un año más de educación, y cada año de educación
vale 25000. O sea que la comparación simple está mezclando dos cosas de signo opuesto: el
castigo de 60000 y la ventaja educativa de 25000. La diferencia de medias recupera la suma de
las dos, no el coeficiente.

Es sesgo por variable omitida, que ya conocen, y acá se puede ver directamente porque sabemos
cuál es la verdad. Es también la razón por la que la regresión existe. Volvemos sobre esto el 8
de octubre.
